# Experiment-1: Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employee_information_100.csv")
```

##Experiment 1.1 — Average salary of each department

```python
avg_salary = df.groupby("Department")["Salary"].mean()

avg_salary.plot(kind="bar")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

##Experiment 1.2 — Number of employees in each department

```python
count = df["Department"].value_counts()
print(count)
```

##Experiment 1.3 — Percentage of male and female employees

```python
gender = df["Gender"].value_counts()

gender.plot(kind="pie", autopct="%1.1f%%")
plt.title("Male and Female Employees")
plt.ylabel("")
plt.show()
```

##Experiment 1.4 — Histogram of salary distribution

```python
plt.hist(df["Salary"], bins=10)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

##Experiment 1.5 — Experience vs Salary

```python
plt.scatter(df["Experience_Years"], df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

##Experiment 1.6 — Top 10 employees with highest salary

```python
top10 = df.sort_values("Salary", ascending=False).head(10)
print(top10[["Name", "Department", "Salary"]])
```

##Experiment 1.7 — Highest salary in every department

```python
highest = df.groupby("Department")["Salary"].max()
print(highest)
```

##Experiment 1.8 — Employees earning more than overall average salary

```python
average = df["Salary"].mean()
result = df[df["Salary"] > average]
print(result[["Name", "Department", "Salary"]])
```

##Experiment 1.9 — Average experience in each department

```python
avg_exp = df.groupby("Department")["Experience_Years"].mean()
print(avg_exp)
```

##Experiment 1.10 — Distribution of employee ages

```python
plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
