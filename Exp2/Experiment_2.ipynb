#2.2 Word Tokenization and Sentence Tokenization

```python
data= """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. NLP is used in many real-world applications, such as chatbots, machine translation, sentiment analysis, and text summarization.

Text tokenization is one of the first steps in NLP. Sentence tokenization divides a paragraph into individual sentences. Word tokenization divides each sentence into individual words and tokens.

For example, a chatbot receives thousands of messages every day. The system processes these messages and identifies useful information. This information can then be used to provide better responses to users."""
```

```python
import nltk

nltk.download('punkt_tab')
```

```python
sentences = nltk.sent_tokenize(data)

print("Sentences:")
for sentence in sentences:
    print(sentence)

words = nltk.word_tokenize(data)

print("\nWords:")
print(words)
```

#2.3 Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
info = """Natural Language Processing is an important field of Artificial Intelligence.
It is used in many applications such as chatbots, machine translation, sentiment analysis, and text summarization.

The main purpose of NLP is to enable computers to understand and process human language.
A computer can analyze a large amount of text in a short period of time.
The removal of stop words is one of the common steps in text preprocessing.

Stop words are commonly occurring words such as the, is, a, an, in, of, and, to, for, on, with, from, by, this, that, and it.
Removing these words can reduce the size of the text and improve the efficiency of some NLP applications."""

stop_words = ["is", "the", "a", "an", "in", "of", "and", "to", "it"]

tokens = info.lower().split()

print("Original Tokens:")
print(tokens)

found_stop_words = []

for word in tokens:
    if word in stop_words:
        found_stop_words.append(word)

filtered_tokens = []

for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)

print("\nStop Words:")
print(found_stop_words)

print("\nFiltered Tokens:")
print(filtered_tokens)
```
