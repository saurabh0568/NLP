### 5.1. Implement subword-level tokenization using a BPE-based tokenizer.

A. Using a pretrained model

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

text = open("input_sub_word_data.txt", encoding="utf-8").read()
tokens = tokenizer.tokenize(text)
ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens)
print("Token IDs:", ids)
```

B. Without a pretrained model

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers

text = open("input_sub_word_data.txt", encoding="utf-8").read()

tokenizer = Tokenizer(models.BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

trainer = trainers.BpeTrainer(
    vocab_size=500,
    special_tokens=["[UNK]"]
)

tokenizer.train_from_iterator([text], trainer)

output = tokenizer.encode(text)

print("Tokens:", output.tokens)
print("Token IDs:", output.ids)
```

 ### 5.2. Implement subword-level tokenization using a SentencePiece tokenizer.

A. Using a pretrained model

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")

text = open("input_sub_word_data.txt", encoding="utf-8").read()

tokens = tokenizer.tokenize(text)
ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens)
print("Token IDs:", ids)
```

B. Without a pretrained model

```python
import sentencepiece as spm

text = open("input_sub_word_data.txt", encoding="utf-8").read()

with open("train.txt", "w", encoding="utf-8") as f:
    f.write(text)

spm.SentencePieceTrainer.train(
    input="train.txt",
    model_prefix="mymodel",
    vocab_size=500,
    model_type="bpe",
    hard_vocab_limit=False
)

sp = spm.SentencePieceProcessor(model_file="mymodel.model")
tokens = sp.encode(text, out_type=str)
ids = sp.encode(text, out_type=int)

print("Tokens:", tokens)
print("Token IDs:", ids)
```

###  5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
                            
                   "The young student is reading an interesting book in the library."   


A. Using spaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = "The young student is reading an interesting book in the library."

doc = nlp(text)
seen = set()

for token in doc:
    if token.text not in seen and not token.is_space:
        seen.add(token.text)
        print(token.text, token.pos_, spacy.explain(token.pos_))
```

B. Using NLTK

```python
import nltk
from nltk import word_tokenize, pos_tag

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("tagsets_json")

text = "The young student is reading an interesting book in the library."

tags = pos_tag(word_tokenize(text))
seen = set()

for word, tag in tags:
    if word not in seen:
        seen.add(word)
        print(word, tag)
        nltk.help.upenn_tagset(tag)
```

###  5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.

```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

text = open("input_sub_word_data.txt", encoding="utf-8").read()
doc = nlp(text)

words = [t.text.lower() for t in doc if t.is_alpha]
freq = Counter(words)
seen = set()

for token in doc:
    word = token.text.lower()

    if token.is_alpha and word not in seen:
        seen.add(word)
        print(word, token.pos_, spacy.explain(token.pos_), freq[word])
```

### 5.5 Write a Python program to train a Word2Vec CBOW model using the given text corpus. Display the vector representation of a selected word and find the three most similar words.
Input corpus
Create a file named corpus.txt:
Natural language processing is interesting
Natural language processing is useful
I love machine learning
Machine learning is useful
Deep learning is a part of machine learning
Artificial intelligence is changing the world
Natural language processing uses machine learning
Deep learning and artificial intelligence are related
I study natural language processing

```python
from gensim.models import Word2Vec

with open("corpus.txt", encoding="utf-8") as f:
    sentences = [line.lower().split() for line in f if line.strip()]

model = Word2Vec(
    sentences, vector_size=50, window=3,
    min_count=1, sg=0, workers=1, seed=42
)

word = "machine"

print("Vector:", model.wv[word])
print("3 similar words:", model.wv.most_similar(word, topn=3))
```

### 5.7 Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.

Calculate similarity for:
machine and learning
natural and language
deep and learning


```python
from gensim.models import Word2Vec

with open("corpus.txt", encoding="utf-8") as f:
    sentences = [line.lower().split() for line in f if line.strip()]

model = Word2Vec(
    sentences, vector_size=50, window=3,
    min_count=1, sg=0, workers=1, seed=42
)

pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

for w1, w2 in pairs:
    score = model.wv.similarity(w1, w2)
    print(w1, "-", w2, ":", round(score, 4))
```
